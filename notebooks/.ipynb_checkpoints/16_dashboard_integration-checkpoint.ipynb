```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

```python
master = pd.read_csv(
    "C:/Users/chowd/Zoonotic-Spillover-Predictor/data/processed/master_features.csv"
)

master.head()
```

```python
print("Districts:")
print(sorted(master["District"].unique()))

print("\nYears:")
print(sorted(master["Year"].unique()))
```

```python
dashboard_summary = (
    master
    .groupby(["State", "District"])
    .agg(
        Mean_Temperature=("Temperature", "mean"),
        Mean_Rainfall=("Rainfall", "mean"),
        Mean_NDVI=("NDVI", "mean"),
        Forest_Loss=("Forest_Loss", "mean"),
        Elevation=("Elevation", "mean")
    )
    .reset_index()
)

dashboard_summary.head()
```

```python
temperature_trend = (
    master
    .groupby("Year")["Temperature"]
    .mean()
    .reset_index()
)

fig = px.line(
    temperature_trend,
    x="Year",
    y="Temperature",
    markers=True,
    title="Average Temperature Trend"
)

fig.update_layout(
    xaxis_title="Year",
    yaxis_title="Temperature (°C)"
)

fig.show()
```

```python
ndvi_trend = (
    master
    .groupby("Year")["NDVI"]
    .mean()
    .reset_index()
)

fig = px.line(
    ndvi_trend,
    x="Year",
    y="NDVI",
    markers=True,
    title="Average Vegetation Health Trend"
)

fig.update_layout(
    xaxis_title="Year",
    yaxis_title="Mean NDVI"
)

fig.show()
```

```python
forest_summary = (
    master
    .groupby("District")["Forest_Loss"]
    .mean()
    .reset_index()
    .sort_values("Forest_Loss", ascending=False)
)

fig = px.bar(
    forest_summary,
    x="District",
    y="Forest_Loss",
    title="Forest Loss by District"
)

fig.update_layout(
    xaxis_title="District",
    yaxis_title="Forest Loss"
)

fig.show()
```

```python
master["Environmental_Risk_Score"] = (
    0.30 * master["Temperature"].rank(pct=True)
    + 0.25 * master["Rainfall"].rank(pct=True)
    + 0.20 * (1 - master["NDVI"].rank(pct=True))
    + 0.25 * master["Forest_Loss"].rank(pct=True)
)

master["Environmental_Risk_Score"] = (
    master["Environmental_Risk_Score"] * 100
)
```

```python
def classify_risk(score):

    if score < 33:
        return "Low Risk"

    elif score < 66:
        return "Moderate Risk"

    else:
        return "High Risk"


master["Environmental_Risk_Level"] = (
    master["Environmental_Risk_Score"]
    .apply(classify_risk)
)
```

```python
master[
    [
        "District",
        "Year",
        "Environmental_Risk_Score",
        "Environmental_Risk_Level"
    ]
].head(20)
```

```python
master.to_csv(
    "C:/Users/chowd/Zoonotic-Spillover-Predictor/data/processed/dashboard_environment_data.csv",
    index=False
)

print("Dashboard dataset saved successfully.")
```
