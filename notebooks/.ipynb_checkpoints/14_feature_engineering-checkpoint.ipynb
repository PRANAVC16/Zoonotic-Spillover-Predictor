```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder
```

```python
master = pd.read_csv(
    "../data/processed/master_environment_dataset.csv"
)

master.head()
```

```python
print(master.shape)

master.info()

master.describe()
```

```python
numeric_cols = master.select_dtypes(include=["number"]).columns

imputer = SimpleImputer(strategy="median")

master[numeric_cols] = imputer.fit_transform(master[numeric_cols])
```

```python
print(master.isnull().sum())
```

```python
state_encoder = LabelEncoder()
district_encoder = LabelEncoder()

master["State_ID"] = state_encoder.fit_transform(master["State"])
master["District_ID"] = district_encoder.fit_transform(master["District"])
```

```python
master["Temp_Range"] = (
    master["Max_Temperature"]
    - master["Min_Temperature"]
)

master["Rainfall_per_Elevation"] = (
    master["Rainfall"]
    / (master["Elevation"] + 1)
)

master["Vegetation_Forest_Index"] = (
    master["NDVI"]
    * (1 - master["Forest_Loss"])
)
```

```python
master.describe()
```

```python
scaler = StandardScaler()

columns_to_scale = [
    "Temperature",
    "Rainfall",
    "NDVI",
    "Forest_Loss",
    "Elevation"
]

scaled = scaler.fit_transform(master[columns_to_scale])

scaled_df = pd.DataFrame(
    scaled,
    columns=[c + "_Scaled" for c in columns_to_scale]
)

master = pd.concat(
    [master, scaled_df],
    axis=1
)
```

```python
correlation = master.select_dtypes(
    include=np.number
).corr()

correlation
```

```python
print(master.describe())

print(master.columns.tolist())
```

```python
master.to_csv(
    "../data/processed/master_features.csv",
    index=False
)

print("Feature engineering complete!")
```

```python
features = pd.read_csv(
    "../data/processed/master_features.csv"
)

print("=" * 70)
print("FEATURE DATASET QA")
print("=" * 70)

print("Shape:", features.shape)

required = [
    "Temperature",
    "Rainfall",
    "NDVI",
    "Temp_Range",
    "Forest_Loss",
    "Elevation"
]

print("\nMissing final modelling features:")
print(
    features[required]
    .isna()
    .sum()
)

print("\nVariance:")
print(
    features[required]
    .var()
    .round(6)
)

print("\nNDVI unique values:")
print(
    features["NDVI"].nunique()
)
```
