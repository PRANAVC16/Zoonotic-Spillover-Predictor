```python
import pandas as pd

districts = pd.read_csv("../data/raw/outbreaks/district_master.csv")

years = list(range(2015, 2025))

rows = []

for _, row in districts.iterrows():
    for year in years:
        rows.append({
            "State": row["State"],
            "District": row["District"],
            "Latitude": row["Latitude"],
            "Longitude": row["Longitude"],
            "Year": year
        })

environment_df = pd.DataFrame(rows)

environment_df.head()
```

```python
print(environment_df.shape)
```

```python
environment_df["Temperature"] = None
environment_df["Rainfall"] = None
environment_df["Humidity"] = None
environment_df["NDVI"] = None
environment_df["Forest_Loss"] = None
environment_df["Elevation"] = None

environment_df.head()
```

```python
environment_df.to_csv(
    "../data/interim/environmental_dataset.csv",
    index=False
)

print("Environmental dataset created successfully!")
```
