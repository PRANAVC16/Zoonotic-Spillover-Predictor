```python
correlation_features = [
    "Temperature",
    "Min_Temperature",
    "Max_Temperature",
    "Rainfall",
    "NDVI",
    "Forest_Loss",
    "Elevation",
    "Temp_Range"
]

corr = df[correlation_features].corr()

corr
```

```python
fig, ax = plt.subplots(figsize=(9, 7))

im = ax.imshow(
    corr.values,
    aspect="auto"
)

ax.set_xticks(
    range(len(correlation_features))
)

ax.set_yticks(
    range(len(correlation_features))
)

ax.set_xticklabels(
    correlation_features,
    rotation=45,
    ha="right"
)

ax.set_yticklabels(
    correlation_features
)

for i in range(len(correlation_features)):
    for j in range(len(correlation_features)):

        ax.text(
            j,
            i,
            f"{corr.iloc[i, j]:.2f}",
            ha="center",
            va="center",
            fontsize=8
        )

fig.colorbar(
    im,
    ax=ax,
    label="Pearson Correlation"
)

ax.set_title(
    "Correlation Matrix of Environmental Variables"
)

plt.tight_layout()

plt.savefig(
    "../paper/figures/figure_correlation_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
profile_features = [
    "Temperature",
    "Rainfall",
    "NDVI",
    "Forest_Loss",
    "Elevation"
]

district_profile = (
    df.groupby("District")[profile_features]
    .mean()
)

normalized_profile = (
    district_profile - district_profile.mean()
) / district_profile.std()

normalized_profile
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


model_results = pd.read_csv(
    "../paper/tables/table_model_comparison_logo.csv"
)

temporal_results = pd.read_csv(
    "../paper/tables/table_temporal_validation.csv"
)

ablation_results = pd.read_csv(
    "../paper/tables/table_feature_set_ablation.csv"
)

oof_predictions = pd.read_csv(
    "../data/processed/model_oof_predictions_logo.csv"
)

feature_importance = pd.read_csv(
    "../paper/tables/table_final_model_feature_importance.csv"
)

print("All final ML outputs loaded.")

display(model_results)
display(temporal_results)
```

```python
model_order = [
    "Logistic Regression",
    "Decision Tree",
    "Random Forest",
    "XGBoost"
]

model_results_ml = model_results[
    model_results["Model"].isin(model_order)
].copy()

model_results_ml["Model"] = pd.Categorical(
    model_results_ml["Model"],
    categories=model_order,
    ordered=True
)

model_results_ml = (
    model_results_ml
    .sort_values("Model")
    .reset_index(drop=True)
)

model_results_ml
```

```python
metrics_to_plot = [
    "Balanced_Accuracy",
    "F1",
    "ROC_AUC",
    "PR_AUC"
]

plot_df = model_results_ml[
    ["Model"] + metrics_to_plot
].set_index("Model")

ax = plot_df.plot(
    kind="bar",
    figsize=(10, 6)
)

ax.set_title(
    "Model Performance Under Leave-One-District-Out Validation"
)

ax.set_xlabel("Model")
ax.set_ylabel("Score")

ax.set_ylim(0, 1)

ax.legend(
    title="Metric",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.xticks(
    rotation=25,
    ha="right"
)

plt.tight_layout()

plt.savefig(
    "../paper/figures/figure_spatial_model_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import joblib

from sklearn.base import clone
from sklearn.model_selection import LeaveOneGroupOut
from sklearn.metrics import (
    roc_curve,
    roc_auc_score,
    precision_recall_curve,
    average_precision_score
)

training = pd.read_csv(
    "../data/processed/training_dataset_verified_v2.csv"
)

final_features = [
    "Temperature",
    "Rainfall",
    "NDVI",
    "Temp_Range"
]

X = training[final_features]
y = training["Outbreak_Label"].astype(int)

groups = training["District"]

final_model = joblib.load(
    "../outputs/models/kfd_outbreak_model_v2.pkl"
)

logo = LeaveOneGroupOut()

true_values = []
predicted_probabilities = []
predicted_classes = []

for train_idx, test_idx in logo.split(
    X,
    y,
    groups
):

    model = clone(
        final_model
    )

    model.fit(
        X.iloc[train_idx],
        y.iloc[train_idx]
    )

    prob = model.predict_proba(
        X.iloc[test_idx]
    )[:, 1]

    pred = model.predict(
        X.iloc[test_idx]
    )

    true_values.extend(
        y.iloc[test_idx].tolist()
    )

    predicted_probabilities.extend(
        prob.tolist()
    )

    predicted_classes.extend(
        pred.tolist()
    )
```

```python
fpr, tpr, _ = roc_curve(
    true_values,
    predicted_probabilities
)

final_auc = roc_auc_score(
    true_values,
    predicted_probabilities
)

plt.figure(
    figsize=(7, 6)
)

plt.plot(
    fpr,
    tpr,
    linewidth=2,
    label=f"Dynamic RF (AUC = {final_auc:.3f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random classifier"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "ROC Curve — Dynamic Random Forest\n"
    "Leave-One-District-Out Validation"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    "../paper/figures/figure_final_dynamic_rf_roc.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(
    "Spatial ROC-AUC:",
    round(final_auc, 3)
)
```

```python
precision, recall, _ = (
    precision_recall_curve(
        true_values,
        predicted_probabilities
    )
)

final_pr_auc = (
    average_precision_score(
        true_values,
        predicted_probabilities
    )
)

plt.figure(
    figsize=(7, 6)
)

plt.plot(
    recall,
    precision,
    linewidth=2,
    label=f"Dynamic RF (AP = {final_pr_auc:.3f})"
)

baseline = np.mean(
    true_values
)

plt.axhline(
    baseline,
    linestyle="--",
    label=f"Positive prevalence = {baseline:.2f}"
)

plt.xlabel("Recall")
plt.ylabel("Precision")

plt.title(
    "Precision–Recall Curve — Dynamic Random Forest"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    "../paper/figures/figure_final_dynamic_rf_precision_recall.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay
)

cm = confusion_matrix(
    true_values,
    predicted_classes
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[
        "No Outbreak",
        "Outbreak"
    ]
)

fig, ax = plt.subplots(
    figsize=(6, 5)
)

disp.plot(
    ax=ax,
    values_format="d"
)

ax.set_title(
    "Dynamic Random Forest — Spatial Validation"
)

plt.tight_layout()

plt.savefig(
    "../paper/figures/figure_final_dynamic_rf_confusion_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(cm)
```

```python
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

final_metrics = pd.DataFrame(
    [
        {
            "Model":
                "Dynamic Random Forest",

            "Validation":
                "Leave-One-District-Out",

            "Accuracy":
                accuracy_score(
                    true_values,
                    predicted_classes
                ),

            "Balanced_Accuracy":
                balanced_accuracy_score(
                    true_values,
                    predicted_classes
                ),

            "Precision":
                precision_score(
                    true_values,
                    predicted_classes
                ),

            "Recall":
                recall_score(
                    true_values,
                    predicted_classes
                ),

            "F1":
                f1_score(
                    true_values,
                    predicted_classes
                ),

            "ROC_AUC":
                final_auc,

            "PR_AUC":
                final_pr_auc
        }
    ]
)

display(
    final_metrics.round(3)
)

final_metrics.to_csv(
    "../paper/tables/table_final_dynamic_rf_results.csv",
    index=False
)
```
