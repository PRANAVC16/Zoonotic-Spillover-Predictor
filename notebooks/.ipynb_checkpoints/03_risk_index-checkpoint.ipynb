```python
import pandas as pd
import numpy as np

climate_df = pd.read_csv("../data/processed/nashik_climate_features.csv")

ndvi_df = pd.read_csv("../data/processed/ndvi_stats.csv")

forest_df = pd.read_csv("../data/processed/forest_loss_stats.csv")

print("Datasets loaded successfully!")
```

```python
print(climate_df.head())
print(ndvi_df.head())
print(forest_df.head())
```

```python
temperature = climate_df["Temperature"].mean()

humidity = climate_df["Humidity"].mean()

ndvi_value = ndvi_df["NDVI"].iloc[0]

forest_loss = forest_df["forest_loss_area_km2"].iloc[0]

print("Temperature:", temperature)
print("Humidity:", humidity)
print("NDVI:", ndvi_value)
print("Forest Loss:", forest_loss)
```

```python
temp_norm = temperature / 50
humidity_norm = humidity / 100
ndvi_risk = 1 - ndvi_value
forest_loss_norm = forest_loss / 100

print(temp_norm)
print(humidity_norm)
print(ndvi_risk)
print(forest_loss_norm)
```

```python
risk_index = (
    (0.3 * temp_norm) +
    (0.2 * humidity_norm) +
    (0.3 * ndvi_risk) +
    (0.2 * forest_loss_norm)
)

print("Spillover Risk Index:", risk_index)
```

```python
if risk_index < 0.3:
    risk_level = "Low Risk"

elif risk_index < 0.6:
    risk_level = "Moderate Risk"

else:
    risk_level = "High Risk"

print("Risk Level:", risk_level)
```

```python
final_df = pd.DataFrame({
    "Temperature": [temperature],
    "Humidity": [humidity],
    "NDVI": [ndvi_value],
    "Forest_Loss": [forest_loss],
    "Risk_Index": [risk_index],
    "Risk_Level": [risk_level]
})

final_df
```

```python
final_df.to_csv("../data/processed/spillover_risk_index.csv", index=False)

print("Spillover Risk Index exported successfully!")
```
