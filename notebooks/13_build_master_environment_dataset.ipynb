```python
import pandas as pd
```

```python
climate = pd.read_csv(
    "../data/processed/climate_western_ghats_2015_2024.csv"
)

ndvi = pd.read_csv(
    "../data/processed/ndvi_western_ghats_2015_2024.csv"
)

forest = pd.read_csv(
    "../data/processed/forest_loss_western_ghats.csv"
)

elevation = pd.read_csv(
    "../data/processed/elevation_western_ghats.csv"
)
print("Datasets Loaded.")
```

```python
datasets = {
    "Climate": climate,
    "NDVI": ndvi,
    "Forest": forest,
    "Elevation": elevation
}

for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print("=" * 60)

    print("Shape:", df.shape)
    print("\nColumns:")
    print(df.columns.tolist())

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicates:", df.duplicated().sum())

    print()
```

```python
master = climate.merge(
    forest,
    on=["State", "District"],
    how="left"
)

print(master.shape)
master.head()
```

```python
print(master.isnull().sum())
```

```python
master = master.merge(
    elevation,
    on=["State", "District"],
    how="left"
)

print(master.shape)

master.head()
```

```python
print(master.isnull().sum())
```

```python
master = master.merge(
    ndvi,
    on=["State", "District", "Year"],
    how="left"
)

print(master.shape)

master.head()
```

```python
print("=" * 60)
print("MASTER DATASET")
print("=" * 60)

print("\nShape:")
print(master.shape)

print("\nColumns:")
print(master.columns.tolist())

print("\nMissing Values:")
print(master.isnull().sum())

print("\nDuplicate Rows:")
print(master.duplicated().sum())

print("\nData Types:")
print(master.dtypes)
```

```python
master.describe(include="all")
```

```python
master.to_csv(
    "../data/processed/master_environment_dataset.csv",
    index=False
)

print("Master Environmental Dataset Saved Successfully!")
```
