```python
import pandas as pd
import matplotlib.pyplot as plt

risk_df = pd.read_csv("../data/processed/spillover_risk_index.csv")

risk_df
```

```python
plt.figure(figsize=(6,4))

plt.bar(
    ["Risk Index"],
    risk_df["Risk_Index"],
)

plt.ylabel("Risk Score")
plt.title("Zoonotic Spillover Risk Index")

plt.ylim(0,1)

plt.show()
```

```python
temperature = risk_df["Temperature"].iloc[0]
humidity = risk_df["Humidity"].iloc[0]
ndvi_value = risk_df["NDVI"].iloc[0]
forest_loss = risk_df["Forest_Loss"].iloc[0]

temp_norm = temperature / 50
humidity_norm = humidity / 100
ndvi_risk = 1 - ndvi_value
forest_loss_norm = forest_loss / 100

print("Values prepared!")
```

```python
features = [
    "Temperature",
    "Humidity",
    "NDVI Risk",
    "Forest Loss"
]

values = [
    temp_norm,
    humidity_norm,
    ndvi_risk,
    forest_loss_norm
]

plt.figure(figsize=(8,5))

plt.bar(features, values)

plt.ylabel("Normalized Contribution")

plt.title("Environmental Drivers of Spillover Risk")

plt.show()
```
