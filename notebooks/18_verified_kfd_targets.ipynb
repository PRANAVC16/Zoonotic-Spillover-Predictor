```python
import pandas as pd
import numpy as np
import os
```

```python
season_years = [2015, 2016, 2017, 2018]

kfd_counts = {

    ("Karnataka", "Shivamogga"):
        [22, 62, 15, 81],

    ("Karnataka", "Uttara Kannada"):
        [1, 0, 0, 0],

    ("Karnataka", "Chikkamagaluru"):
        [0, 0, 0, 0],

    ("Karnataka", "Kodagu"):
        [0, 0, 0, 0],

    ("Kerala", "Wayanad"):
        [7, 0, 0, 5],

    ("Kerala", "Kannur"):
        [0, 0, 0, 0],

    ("Goa", "North Goa"):
        [191, 79, 58, 16],

    ("Goa", "South Goa"):
        [0, 0, 0, 1],

    ("Maharashtra", "Sindhudurg"):
        [25, 87, 35, 34]
}
```

```python
source_pages = {
    "Goa": "52-53",
    "Karnataka": "53-54",
    "Kerala": "54",
    "Maharashtra": "54-55"
}

rows = []

for (state, district), values in kfd_counts.items():

    for year, cases in zip(
        season_years,
        values
    ):

        rows.append({

            "State": state,

            "District": district,

            # Start year of KFD surveillance season
            "Year": year,

            "KFD_Season":
                f"{year}-{str(year + 1)[-2:]}",

            "KFD_AFI_Cases":
                cases,

            # Primary ML target
            "Outbreak_Label":
                int(cases > 0),

            "Target_Definition":
                "Reported KFD occurrence in NCDC AFI surveillance annexure",

            "Case_Definition":
                "KFD cases listed in AFI surveillance affected-village annexure",

            "Source_ID":
                "NCDC_KFD_AFI_2014_2019",

            "Source_Document":
                "Kyasanur Forest Disease: A Public Health Concern",

            "Source_Page":
                source_pages[state],

            "Evidence_Status":
                "Verified",

            "Zero_Basis":
                (
                    "Case/affected village not listed for this "
                    "district-season in NCDC AFI annexure"
                    if cases == 0
                    else ""
                )
        })


labels = pd.DataFrame(rows)

labels = labels.sort_values(
    ["Year", "State", "District"]
).reset_index(drop=True)

labels.head(20)
```

```python
print("=" * 60)
print("VERIFIED KFD TARGET DATASET")
print("=" * 60)

print("\nShape:")
print(labels.shape)

print("\nSeasons:")
print(
    labels[
        ["Year", "KFD_Season"]
    ].drop_duplicates()
)

print("\nClass Distribution:")
print(
    labels["Outbreak_Label"]
    .value_counts()
    .sort_index()
)

print("\nCases by District:")
print(
    labels.groupby("District")[
        "KFD_AFI_Cases"
    ].sum().sort_values(
        ascending=False
    )
)
```

```python
target_matrix = labels.pivot(
    index="District",
    columns="Year",
    values="Outbreak_Label"
)

target_matrix
```

```python
case_matrix = labels.pivot(
    index="District",
    columns="Year",
    values="KFD_AFI_Cases"
)

case_matrix
```

```python
target_path = (
    "../data/raw/outbreaks/"
    "kfd_verified_targets_v1.csv"
)

labels.to_csv(
    target_path,
    index=False
)

print(
    "Saved:",
    target_path
)
```

```python
environment = pd.read_csv(
    "../data/processed/master_features.csv"
)

environment["Year"] = (
    environment["Year"]
    .astype(int)
)

labels["Year"] = (
    labels["Year"]
    .astype(int)
)

print(
    "Environmental shape:",
    environment.shape
)

print(
    "Target shape:",
    labels.shape
)
```

```python
training = environment.merge(

    labels,

    on=[
        "State",
        "District",
        "Year"
    ],

    how="inner",

    validate="many_to_one"
)

print(
    "Training dataset shape:",
    training.shape
)

training.head()
```

```python
print("=" * 60)
print("TRAINING DATA AUDIT")
print("=" * 60)

print("\nShape:")
print(training.shape)

print("\nClass balance:")
print(
    training[
        "Outbreak_Label"
    ].value_counts()
)

print("\nClass percentage:")
print(
    training[
        "Outbreak_Label"
    ].value_counts(
        normalize=True
    ) * 100
)

print("\nRows by district:")
print(
    training.groupby(
        "District"
    ).size()
)

print("\nYears:")
print(
    sorted(
        training["Year"].unique()
    )
)

print("\nMissing values:")
print(
    training.isnull()
    .sum()
    .sort_values(
        ascending=False
    )
    .head(10)
)
```

```python
training.to_csv(
    "../data/processed/"
    "training_dataset_verified_v1.csv",
    index=False
)

print(
    "Verified training dataset V1 saved."
)
```

```python
# Official Karnataka KFD district data, 2020-2024
# Source: Karnataka Health & Family Welfare Annual Report 2024-25

recent_kfd = {

    ("Karnataka", "Shivamogga"): {
        "cases":  [185, 15, 35, 15, 63],
        "deaths": [4,   0,  1,  0,  1]
    },

    ("Karnataka", "Uttara Kannada"): {
        "cases":  [89, 5, 14, 0, 113],
        "deaths": [1,  0,  1, 0,   9]
    },

    ("Karnataka", "Chikkamagaluru"): {
        "cases":  [12, 3, 0, 2, 126],
        "deaths": [0,  0, 0, 0,   4]
    },

    ("Karnataka", "Kodagu"): {
        "cases":  [0, 0, 0, 0, 0],
        "deaths": [0, 0, 0, 0, 0]
    }
}

recent_years = [
    2020,
    2021,
    2022,
    2023,
    2024
]
```

```python
recent_rows = []

for (state, district), values in recent_kfd.items():

    for year, cases, deaths in zip(
        recent_years,
        values["cases"],
        values["deaths"]
    ):

        recent_rows.append({

            "State": state,
            "District": district,
            "Year": year,

            "Target_Period": str(year),

            "Target_Period_Type":
                "Calendar Year",

            "KFD_Cases": cases,

            "Deaths": deaths,

            "Outbreak_Label":
                int(cases > 0),

            "Source_ID":
                "KARNATAKA_HFW_2024_25",

            "Source_Document":
                "Karnataka Health and Family Welfare Annual Report 2024-25",

            "Evidence_Status":
                "Verified"
        })


recent_labels = pd.DataFrame(
    recent_rows
)

recent_labels
```

```python
print("Shape:")
print(recent_labels.shape)

print("\nOutbreak Distribution:")
print(
    recent_labels[
        "Outbreak_Label"
    ].value_counts()
)

print("\nCases by District:")
print(
    recent_labels
    .groupby("District")["KFD_Cases"]
    .sum()
    .sort_values(
        ascending=False
    )
)
```

```python
historical_labels = labels.copy()

historical_labels[
    "KFD_Cases"
] = historical_labels[
    "KFD_AFI_Cases"
]

historical_labels[
    "Deaths"
] = pd.NA

historical_labels[
    "Target_Period"
] = historical_labels[
    "KFD_Season"
]

historical_labels[
    "Target_Period_Type"
] = "NCDC Surveillance Season"
```

```python
target_columns = [

    "State",
    "District",
    "Year",

    "Target_Period",
    "Target_Period_Type",

    "KFD_Cases",
    "Deaths",

    "Outbreak_Label",

    "Source_ID",
    "Source_Document",

    "Evidence_Status"
]
```

```python
historical_clean = historical_labels[
    target_columns
].copy()

recent_clean = recent_labels[
    target_columns
].copy()
```

```python
targets_v2 = pd.concat(
    [
        historical_clean,
        recent_clean
    ],
    ignore_index=True
)

targets_v2 = (
    targets_v2
    .sort_values(
        [
            "Year",
            "State",
            "District"
        ]
    )
    .reset_index(
        drop=True
    )
)

print(
    "Verified target rows:",
    targets_v2.shape
)

targets_v2.head()
```

```python
duplicates = targets_v2.duplicated(
    subset=[
        "State",
        "District",
        "Year"
    ],
    keep=False
)

print(
    "Duplicate district-year records:",
    duplicates.sum()
)

targets_v2[
    duplicates
]
```

```python
targets_v2.to_csv(

    "../data/raw/outbreaks/"
    "kfd_verified_targets_v2.csv",

    index=False
)

print(
    "Verified KFD target dataset V2 saved."
)
```

```python
environment = pd.read_csv(
    "../data/processed/master_features.csv"
)

environment["Year"] = (
    environment["Year"]
    .astype(int)
)

targets_v2["Year"] = (
    targets_v2["Year"]
    .astype(int)
)


training_v2 = environment.merge(

    targets_v2,

    on=[
        "State",
        "District",
        "Year"
    ],

    how="inner",

    validate="one_to_one"
)

print(
    "Training V2 shape:",
    training_v2.shape
)
```

```python
print("=" * 65)
print("FINAL VERIFIED TRAINING DATASET V2")
print("=" * 65)

print("\nShape:")
print(training_v2.shape)

print("\nClass Balance:")
print(
    training_v2[
        "Outbreak_Label"
    ].value_counts()
)

print("\nClass Percentage:")
print(
    training_v2[
        "Outbreak_Label"
    ].value_counts(
        normalize=True
    ) * 100
)

print("\nRows by District:")
print(
    training_v2
    .groupby("District")
    .size()
)

print("\nRows by Year:")
print(
    training_v2
    .groupby("Year")
    .size()
)

print("\nCases by District:")
print(
    training_v2
    .groupby("District")[
        "KFD_Cases"
    ]
    .sum()
    .sort_values(
        ascending=False
    )
)
```

```python
training_v2.to_csv(

    "../data/processed/"
    "training_dataset_verified_v2.csv",

    index=False
)

print(
    "Final verified training dataset V2 saved."
)
```
